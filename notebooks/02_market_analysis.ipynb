# Used Car Market Intelligence — Market Analysis

Descriptive fleet/resale analysis on the validation output.

```python
from pathlib import Path
import pandas as pd
ROOT = Path('..').resolve()
df = pd.read_csv(ROOT / 'data/processed/used_cars_clean.csv')
```

```python
df.groupby('brand').agg(listings=('price','size'), median_price=('price','median'), median_mileage=('mileage','median')).sort_values('median_price', ascending=False)
```

```python
df.groupby('vehicle_age').agg(listings=('price','size'), median_price=('price','median'), median_mileage=('mileage','median')).round(1)
```

```python
bands = pd.cut(df['mileage'], [-1,15000,30000,50000,float('inf')], labels=['<15k','15k-30k','30k-50k','50k+'])
df.assign(mileage_band=bands).groupby('mileage_band', observed=True).agg(listings=('price','size'), avg_price=('price','mean'), avg_vehicle_age=('vehicle_age','mean')).round(1)
```

Price, mileage and vehicle age move together in this sample. The analysis therefore treats these as associated market signals rather than causal depreciation estimates.
